```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# -----------------------------
# Load data
# -----------------------------

PROJECT_ROOT = Path(__file__).resolve().parents[1]

df = pd.read_csv(
    PROJECT_ROOT / "data" / "processed" / "dataset.csv",
    index_col=0,
    parse_dates=True
)

# -----------------------------
# Plot realized volatility
# -----------------------------

plt.figure(figsize=(14,6))

plt.plot(
    df.index,
    df["volatility"],
    linewidth=1.2
)

plt.title("Monthly Realized S&P 500 Volatility")
plt.xlabel("Year")
plt.ylabel("Volatility")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```
